```python
from PIL import Image
import pytesseract
import requests
from io import BytesIO
from PIL import Image, ImageEnhance
# Tesseract 경로 설정 (Windows에서 Tesseract가 설치된 경로를 입력해야 합니다)
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"  # 경로 수정 필요

# 이미지 URL에서 가져오기
image_url = "https://www.chungmafishing.com/_core/upload/temp_upload/2019chungma/20250417003713_ogsmzkcv.jpg"

# 예외 처리 추가
try:
    response = requests.get(image_url)
    response.raise_for_status()  # 요청이 성공했는지 확인
    img = Image.open(BytesIO(response.content))
    img = img.convert('L')  # 흑백 변환
    enhancer = ImageEnhance.Contrast(img)
    img = enhancer.enhance(2)  # 대비 조정
    # Tesseract를 사용하여 이미지에서 텍스트 추출 (한글을 위한 'kor' 언어 데이터 필요)
    text = pytesseract.image_to_string(img, lang='kor', config='--psm 1')

    # 추출된 텍스트 출력
    print(text)
except requests.exceptions.RequestException as e:
    print(f"이미지 다운로드 실패: {e}")
except Exception as e:
    print(f"텍스트 추출 실패: {e}")
```
