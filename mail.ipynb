```python
import smtplib
from email.mime.text import MIMEText
from email.mime.multipart import MIMEMultipart
import urllib.parse  # 비밀번호 인코딩

# SMTP 서버 설정
smtp_server = "smtp.naver.com"
smtp_port = 465  # TLS 포트
username = 'danakkafishing@naver.com'
password = 'kms2211@@'  # 비밀번호에 : 포함

# 이메일 메시지 구성
message = MIMEMultipart()
message["From"] = username
message["To"] = "kkmhss702@naver.com"
message["Subject"] = "테스트 이메일"

body = "이것은 SMTP를 통해 보낸 테스트 이메일입니다."
message.attach(MIMEText(body, "plain"))

try:
    # SSL을 사용하여 SMTP 서버에 직접 연결
    with smtplib.SMTP_SSL(smtp_server, smtp_port) as server:
        server.ehlo()  # 서버 확인
        
        # 로그인 (앱 비밀번호 사용 권장)
        server.login(username, password)
        print("✅ 로그인 성공 (SSL)")
        
        # 이메일 전송
        server.sendmail(username, "kkmhss702@naver.com", message.as_string())
        print("✅ 이메일 전송 성공 (SSL)")

except smtplib.SMTPAuthenticationError as e:
    print(f"❌ 이메일 인증 오류: {e}")
except smtplib.SMTPException as e:
    print(f"❌ 이메일 전송 오류: {e}")
except Exception as e:
    print(f"❌ 알 수 없는 오류: {e}")
```
